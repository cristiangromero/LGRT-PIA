```python
# Ejercicio A1
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

vocabulario = [
    'perro', 'gato', 'elefante', 'casa',
    'coche', 'avión', 'bicicleta', 'león'
]

# Crear representación One-Hot
one_hot = np.eye(len(vocabulario))

print("Representaciones One-Hot:")
for palabra, vector in zip(vocabulario, one_hot):
    print(f"{palabra}: {vector.astype(int)}")

# Calcular matriz de similitud coseno
matriz_similitud = cosine_similarity(one_hot)

print("\nMatriz de similitud:")
print(np.round(matriz_similitud, 2))
```

```python
# Ejercicio A2
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

vocabulario = [
    'perro', 'gato', 'elefante', 'casa',
    'coche', 'avión', 'bicicleta', 'león'
]

caracteristicas = [
    'es_animal',
    'es_domestico',
    'es_felino',
    'tiene_motor',
    'es_transporte',
    'es_grande',
    'vuela'
]

embeddings = np.array([
    [0.90,  0.80, -0.20, -0.80, -0.70,  0.60, -0.90],  # perro
    [0.80,  0.90,  0.90, -0.80, -0.70,  0.20, -0.90],  # gato
    [0.90,  0.10, -0.70, -0.90, -0.80,  1.00, -0.80],  # elefante
    [-0.80, 0.80, -1.00, -1.00, -1.00,  0.30, -1.00],  # casa
    [-1.00,-1.00, -1.00,  0.90,  0.90,  0.30, -1.00],  # coche
    [-1.00,-1.00, -1.00,  1.00,  1.00,  0.50,  1.00],  # avión
    [-1.00,-1.00, -1.00, -0.30,  0.90, -0.10, -1.00],  # bicicleta
    [0.95, -0.70,  0.95, -1.00, -1.00,  0.90, -0.80]   # león
])

df_embeddings = pd.DataFrame(
    embeddings,
    index=vocabulario,
    columns=caracteristicas
)

print("Embeddings:")
print(df_embeddings)

# Matriz de similitud coseno
matriz_similitud = cosine_similarity(embeddings)

df_similitud = pd.DataFrame(
    matriz_similitud,
    index=vocabulario,
    columns=vocabulario
)

print("\nMatriz de similitud:")
print(df_similitud.round(3))
```

```python
import math
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

CASAS = [
    (50, 100000),
    (80, 160000),
    (120, 240000),
    (90, 180000),
    (60, 120000),
    (150, 300000),
]

N = len(CASAS)


class RegresionLinealGD:
    def __init__(self, peso_inicial):
        self.peso = peso_inicial

    def predecir(self, x, peso=None):
        w = self.peso if peso is None else peso
        return w * x

    def perdida_mse(self, peso=None):
        w = self.peso if peso is None else peso
        return sum((self.predecir(x, w) - y) ** 2 for x, y in CASAS) / N

    def gradiente(self, peso=None):
        w = self.peso if peso is None else peso
        return (2.0 / N) * sum((self.predecir(x, w) - y) * x for x, y in CASAS)

    def entrenar(self, tasa_aprendizaje, epocas, limite=1e15):
        historial_peso = [self.peso]
        historial_perdida = [self.perdida_mse()]

        for epoca in range(1, epocas + 1):
            grad = self.gradiente()
            nuevo_peso = self.peso - tasa_aprendizaje * grad

            if not math.isfinite(nuevo_peso) or abs(nuevo_peso) > limite:
                print(f"  Época {epoca}: |peso| supera {limite:.0e}, se detiene el entrenamiento.")
                break

            self.peso = nuevo_peso
            perdida = self.perdida_mse()
            historial_peso.append(self.peso)
            historial_perdida.append(perdida)
            print(f"  Época {epoca}: peso = {self.peso:.4f}  |  pérdida (MSE) = {perdida:.6g}")

        return historial_peso, historial_perdida


def graficar_convergencia(historiales, archivo):
    plt.figure(figsize=(7.5, 4.8))
    for etiqueta, historial_perdida in historiales.items():
        limite = min(len(historial_perdida), 10)
        plt.plot(range(limite), historial_perdida[:limite], marker="o", label=etiqueta)
    plt.yscale("log")
    plt.xlabel("Época")
    plt.ylabel("Pérdida (MSE, escala logarítmica)")
    plt.title("Convergencia del error por tasa de aprendizaje")
    plt.legend()
    plt.grid(True, which="both", alpha=0.3)
    plt.tight_layout()
    plt.savefig(archivo, dpi=150)
    plt.close()


if __name__ == "__main__":
    tasas_aprendizaje = [0.001, 0.01, 0.1]
    epocas_totales = 80

    historiales_perdida = {}

    for tasa in tasas_aprendizaje:
        print("=" * 70)
        print(f"ENTRENAMIENTO CON TASA DE APRENDIZAJE = {tasa}")
        print("=" * 70)
        modelo = RegresionLinealGD(peso_inicial=1000.0)
        historial_peso, historial_perdida = modelo.entrenar(tasa, epocas_totales)
        historiales_perdida[f"lr={tasa}"] = historial_perdida
        print()
```

```python
# Ejercicio D - Generalización en Clasificación
# 0 = Perro
# 1 = Gato

# Dataset de entrenamiento
animales_entrenamiento = [
    [0.3, 0.4, 1, 0.8],  # Gato doméstico
    [0.2, 0.5, 1, 0.9],  # Gato doméstico 2
    [0.4, 0.3, 1, 0.7],  # Gato doméstico 3
    [0.6, 0.8, 0, 0.3],  # Perro mediano
    [0.7, 0.9, 0, 0.2],  # Perro grande
]

# Respuestas correctas del entrenamiento
etiquetas = [1, 1, 1, 0, 0]

# Casos nuevos que el modelo nunca vio
casos_nuevos = [
    [0.25, 0.45, 1, 0.85],  # Gato persa
    [0.35, 0.4, 1, 0.75],   # Gato siamés
    [0.8, 1.0, 0, 0.1],     # Pastor alemán
    [0.5, 0.2, 1, 0.6],     # Gato manx
]

nombres = [
    "Gato persa",
    "Gato siamés",
    "Pastor alemán",
    "Gato manx"
]


# Valores iniciales del perceptrón
pesos = [0.0, 0.0, 0.0, 0.0]
sesgo = 0.0
tasa_aprendizaje = 0.1
epocas = 20


def predecir(datos):
    suma = sesgo

    for i in range(len(datos)):
        suma += datos[i] * pesos[i]

    if suma >= 0:
        return 1
    else:
        return 0


# Entrenamiento
for epoca in range(epocas):

    for datos, etiqueta_real in zip(
        animales_entrenamiento,
        etiquetas
    ):
        prediccion = predecir(datos)

        error = etiqueta_real - prediccion

        # Ajuste de los pesos
        for i in range(len(pesos)):
            pesos[i] += tasa_aprendizaje * error * datos[i]

        # Ajuste del sesgo
        sesgo += tasa_aprendizaje * error


# Mostrar lo aprendido
print("=== CLASIFICADOR DE PERROS Y GATOS ===")

print("\nPesos aprendidos:")
print(pesos)

print("Sesgo:")
print(sesgo)


# Prueba con casos nunca vistos
print("\n=== CASOS NUEVOS ===")

for nombre, datos in zip(nombres, casos_nuevos):

    resultado = predecir(datos)

    if resultado == 1:
        clasificacion = "Gato"
    else:
        clasificacion = "Perro"

    print(f"{nombre}: {clasificacion}")

    # Pruebas adicionales para analizar los límites de generalización
print("\n=== PRUEBAS DE GENERALIZACIÓN ===")

casos_limite = [
    [0.65, 0.85, 1, 0.8],   # Mezcla características de perro y gato
    [0.25, 0.35, 0, 0.2],   # Mezcla características de gato y perro
]

nombres_limite = [
    "Caso ambiguo 1",
    "Caso ambiguo 2"
]

for nombre, datos in zip(nombres_limite, casos_limite):
    resultado = predecir(datos)

    if resultado == 1:
        clasificacion = "Gato"
    else:
        clasificacion = "Perro"

    print(f"{nombre}: {datos} -> {clasificacion}")
```
