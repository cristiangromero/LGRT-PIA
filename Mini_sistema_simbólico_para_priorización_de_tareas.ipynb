```python
# Hechos iniciales
hechos = {
    ("tipo", "Tarea1", "Incidente"),
    ("impacto", "Tarea1", "Alto"),
    ("prioridad", "Tarea1", "Media"),
    ("estado", "Tarea1", "Pendiente"),

    ("tipo", "Tarea2", "Mejora"),
    ("impacto", "Tarea2", "Medio"),
    ("prioridad", "Tarea2", "Alta"),
    ("estado", "Tarea2", "Pendiente"),

    ("tipo", "Tarea3", "Documentacion"),
    ("impacto", "Tarea3", "Bajo"),
    ("prioridad", "Tarea3", "Baja"),
    ("estado", "Tarea3", "Pendiente")
}


def aplicar_reglas(hechos):
    nuevas_conclusiones = []

    for hecho in hechos.copy():
        if hecho[0] == "tipo":
            tarea = hecho[1]
            tipo = hecho[2]

            # Regla 1
            if (tipo == "Incidente" and
                ("impacto", tarea, "Alto") in hechos):

                conclusion = ("prioridad_ejecucion", tarea, "Urgente")

                if conclusion not in hechos:
                    nuevas_conclusiones.append(conclusion)

            # Regla 2
            if (tipo == "Mejora" and
                ("prioridad", tarea, "Alta") in hechos and
                ("estado", tarea, "Pendiente") in hechos):

                conclusion = ("prioridad_ejecucion", tarea, "Alta")

                if conclusion not in hechos:
                    nuevas_conclusiones.append(conclusion)

    for hecho in hechos.copy():
        if hecho[0] == "prioridad":
            tarea = hecho[1]
            prioridad = hecho[2]

            # Regla 3
            if (prioridad == "Baja" and
                ("impacto", tarea, "Bajo") in hechos):

                conclusion = ("prioridad_ejecucion", tarea, "Normal")

                if conclusion not in hechos:
                    nuevas_conclusiones.append(conclusion)

    return nuevas_conclusiones


# Aplicar el motor de inferencia
conclusiones = aplicar_reglas(hechos)

for conclusion in conclusiones:
    print(conclusion)
```
